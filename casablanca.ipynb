```python
import geopandas as gpd
import osmnx as ox

# 1. Get the administrative boundary of Casablanca
city = "Casablanca, Morocco"
casablanca_boundary = ox.geocode_to_gdf(city)

# Save boundary as GeoJSON
casablanca_boundary.to_file("casablanca_boundary.geojson", driver="GeoJSON")
print("✅ Saved Casablanca city boundary to casablanca_boundary.geojson")

# 2. Get districts (admin_level 8)
districts = ox.geometries_from_place(city, tags={"boundary": "administrative", "admin_level": "8"})
districts = districts[districts.geometry.type.isin(["Polygon", "MultiPolygon"])]

# Save districts as GeoJSON
districts.to_file("casablanca_districts.geojson", driver="GeoJSON")
print("✅ Saved Casablanca districts to casablanca_districts.geojson")

districts[["name", "geometry"]].head()
```
